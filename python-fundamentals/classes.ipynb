# Classes and Objects
- **Beyond Built-ins:** Python lets you define your own data types using `class`.
- **Class:** A blueprint or template for creating objects. Defines attributes (data) and methods (behavior). Convention: `PascalCase` names (`MyClass`).
- **Object (Instance):** A specific item created from a class blueprint. Each object has its own set of attribute values but shares the methods defined by the class. `obj1 = MyClass()`, `obj2 = MyClass()`. `obj1` and `obj2` are distinct objects.

## Defining a Class & `__init__` (The Constructor)
- **`__init__(self, ...)`:** Special method for initialization. `self` is always the first parameter and represents the instance itself. Other parameters receive arguments passed during object creation.
- **Instance Attributes (`self.x = ...`):** Data attached to *this specific object*. Created inside methods (usually `__init__`) using `self.attribute_name = value`.

```python
class ServiceMonitor:
    """ Provides service checks for a single service """
    def __init__(self, service_name, port):
        """Initializes the monitor for a specific service.
        Args:
            service_name (str): The name of the service
            port (int): The port number of the service
        """
        print(f"Initializing monitor for {service_name} on port {port}.")
        self.service = service_name
        self.port = port
        self.is_alive = False

    def check(self):
        """Performs a service check."""
        print(f"METHOD: Checking {self.service} on port {self.port}.")
        self.is_alive = True
        print(f"METHOD: Status for service {self.service}: {"Alive" if self.is_alive else "Down"}")
        return self.is_alive
```

## Creating Instances (Objects)
- **Mechanism:** Call the class name as if it were a function, passing any arguments required by `__init__` (after `self`).
- Python automatically creates the object and passes it as `self` to `__init__`.

```python
nginx_monitor = ServiceMonitor("nginx", 80)
status = nginx_monitor.check()
print(f"Received status: {status}")
```

## Instance Methods: Object Behavior
- **Definition:** Functions defined *inside* a class definition.
- **First Parameter:** Always `self` (by strong convention), allowing the method to access and modify the instance's attributes (`self.attribute_name`).
- **Calling:** Use dot notation on an instance: `instance.method_name(arguments)`. Python automatically passes the instance (`instance`) as the `self` argument.

## Basic Inheritance: Reusing and Extending
- **Concept:** Create a new class (Child/Subclass) that inherits properties (attributes and methods) from an existing class (Parent/Superclass). Promotes code reuse (DRY).
- **Syntax:** `class ChildClassName(ParentClassName):`
- **Inherited Members:** The Child automatically gets all methods and attributes defined in the Parent.
- **Specializing:** The Child can:
  - Add *new* attributes and methods.
  - *Override* parent methods by defining a method with the same name.
- **`super()`:** Inside the Child's methods, use `super().method_name(...)` to explicitly call the Parent's version of a method (very common in `__init__`).

```python
class HttpServiceMonitor(ServiceMonitor):
    """Extends ServiceMonitor to add a HTTP endpoint check"""
    def __init__(self, service_name, port, url):
        super().__init__(service_name, port)
        self.url = url

    def ping(self):
        """Ping url """
        print(f"METHOD: Ping url: {self.url}")

location_monitor = HttpServiceMonitor("localweb", 8080, "http://localhost")
location_monitor.ping()
```

Imaginea completă, explicată pas cu pas: cum legăm clasa-părinte, subclasele (copiii), atributele și metodele între ele.

Gândește-te la o familie din lumea reală:

    Părintele (Vehicul) are trăsături generale pe care le transmite mai departe: toate vehiculele au marcă, model și pot porni.

    Copilul / Subclasa (MasinaElectrica) moștenește tot ce are părintele, dar vine cu dotări proprii (de exemplu, baterie și încărcare).

Conceptele cheie pe scurt

    Atribut: Ce are obiectul (date, variabile: marca, baterie).

    Metodă: Ce face obiectul (acțiuni, funcții: porneste(), incarca()).

    Moștenire (Subclasă): O clasă copil care preia automat totul de la clasa părinte, fără să rescrii codul.

    super(): Modul prin care copilul îi spune părintelui: "Fă tu partea ta de inițializare, că mă ocup eu de restul".

```python
# 1. CLASA PĂRINTE (Clasa de bază)
class Vehicul:
    def __init__(self, marca, model):
        # ATRIBUTE comune tuturor vehiculelor
        self.marca = marca
        self.model = model
        self.pornit = False

    # METODE comune
    def porneste(self):
        self.pornit = True
        print(f"{self.marca} {self.model} a pornit.")

    def claxoneaza(self):
        print("Beep beep!")


# 2. SUBCLASA (Copilul) - moștenește din Vehicul
class MasinaElectrica(Vehicul):
    def __init__(self, marca, model, capacitate_baterie):
        # Pasăm marca și modelul către părinte folosind super()
        super().__init__(marca, model)

        # ATRIBUT SPECIFIC doar pentru mașini electrice
        self.capacitate_baterie = capacitate_baterie
        self.nivel_baterie = 100

    # METODĂ SPECIFICĂ doar acestei subclase
    def incarca(self):
        self.nivel_baterie = 100
        print(f"{self.marca} {self.model} s-a încărcat complet (100%).")

    # SUPRASCRIERE DE METODĂ (Polimorfism):
    # Schimbăm comportamentul metodei din părinte ca să se potrivească unei electrice
    def claxoneaza(self):
        print("Sunet futurist: Vuuuuum!")
```

```python
# Creăm un vehicul simplu (folosește doar clasa părinte)
camion = Vehicul("Volvo", "FH16")
camion.porneste()     # Ieșire: Volvo FH16 a pornit.
camion.claxoneaza()   # Ieșire: Beep beep!

print("-" * 30)

# Creăm o instanță a subclasei
tesla = MasinaElectrica("Tesla", "Model 3", 75)

# 1. Moștenire: folosește metoda 'porneste()' definită în Vehicul!
tesla.porneste()      # Ieșire: Tesla Model 3 a pornit.

# 2. Specificitate: are atribute și metode pe care camionul NU le are
print(f"Baterie: {tesla.capacitate_baterie} kWh")  # Ieșire: Baterie: 75 kWh
tesla.incarca()       # Ieșire: Tesla Model 3 s-a încărcat complet (100%).

# 3. Suprascriere: claxonul a fost modificat doar pentru ea
tesla.claxoneaza()    # Ieșire: Sunet futurist: Vuuuuum!
```

Cont bancar:
Fiecare cont bancar are nevoie de:

    Ce are (Atribute): titular (numele) și sold (banii din cont).

    Ce poate face (Metode): să depună bani (depune), să scoată bani (retrage) și să verifice câți bani mai are (arata_sold).

```python
class ContBancar:
    # 1. Constructor: setează titularul și soldul inițial
    def __init__(self, titular, sold_initial=0):
        self.titular = titular
        self.sold = sold_initial

    # 2. Metodă pentru depunere
    def depune(self, suma):
        self.sold += suma
        print(f"Ai depus {suma} lei. Sold nou: {self.sold} lei.")

    # 3. Metodă pentru retragere (cu verificare să nu iasă pe minus)
    def retrage(self, suma):
        if suma > self.sold:
            print(f"Fonduri insuficiente! Ai doar {self.sold} lei.")
        else:
            self.sold -= suma
            print(f"Ai retras {suma} lei. Sold rămas: {self.sold} lei.")

    # 4. Metodă simplă de afișare
    def arata_sold(self):
        print(f"Contul lui {self.titular} are {self.sold} lei.")


# SUBCLASA (mosteneste din cont bancar)
class ContEconomii(ContBancar):
    def __init__(self, titular, sold_initial=0, dobanda=0.05):
        # 1. super() trimite titularul și soldul către clasa părinte
        super().__init__(titular, sold_initial)

        # 2. Atribut propriu doar pentru economii
        self.dobanda = dobanda

    # 3. Metodă specifică doar pentru economii
    def aplica_dobanda(self):
        castig = self.sold * self.dobanda
        self.sold += castig
        print(f"Dobanda adaugata: {castig:.2f} lei. Sold nou: {self.sold:.2f} lei.")

# Creăm un cont pentru Andrei cu 100 de lei la start
cont_andrei = ContBancar("Andrei", 100)

# Verificăm starea
cont_andrei.arata_sold()  # Contul lui Andrei are 100 lei.

# Depunem 50 lei
cont_andrei.depune(50)    # Ai depus 50 lei. Sold nou: 150 lei.

# Încercăm să retragem prea mult
cont_andrei.retrage(200)  # Fonduri insuficiente! Ai doar 150 lei.

# Retragere validă
cont_andrei.retrage(30)   # Ai retras 30 lei. Sold rămas: 120 lei.

print("-" * 30)

# Creăm un cont de economii pentru Elena cu 1000 lei și 5% dobândă
economii_elena = ContEconomii("Elena", 1000, 0.05)

# 1. Poate folosi metodele din părinte:
economii_elena.arata_sold()     # Contul lui Elena are 1000 lei.
economii_elena.depune(200)       # Ai depus 200 lei. Sold nou: 1200 lei.

# 2. Folosim funcția proprie de dobândă (5% din 1200 = 60 lei):
economii_elena.aplica_dobanda()  # Dobândă adăugată: 60.00 lei. Sold nou: 1260.00 lei.
```

Exercise: Object-Oriented Deployment Manager - Core Logic

You are a senior DevOps engineer designing a new deployment tool. Your first step is to create the core logic for a Deployment class. This class will model a software deployment, encapsulating its state and version history. For this part, you will focus on the "happy path" and assume all inputs are valid.


Problem Description

Your task is to implement the Deployment class in exercise.py. This class must provide a specific public interface to manage a deployment's state and history.


Behavioral Requirements

1. __init__(self, service_name, environment):

    A new deployment object should be initialized with an initial status of 'pending' and no deployed version. It should store the service_name and environment.

    service_name and environment are expected to be strings.

2. deploy(self, new_version):

    A deployment should update the status to 'deployed' and make new_version the current version. The previously current version should be retained internally to support rollbacks.

3. rollback(self):

    A rollback reverts the deployment to the version that was active immediately before the current one.

        A rollback is only possible if there is a historical version to revert to. If not, this method must return False.

        On a successful rollback, the status should be updated to 'rolled_back', and the method should return True. Additionally, the check_status method should now return the version to which the deployment was rolled back.

4. check_status(self):

    This method must return a dictionary with four keys:

        'service_name' and 'environment': The values provided during initialization.

        'status': The current status ('pending', 'deployed', or 'rolled_back').

        'version': The currently active version string. If no version has been deployed, this must be None.

```python
class Deployment:
    """
    Manages the state and version history of a software deployment.
    """

    def __init__(self, service_name: str, environment: str):
        self.service_name = service_name
        self.environment = environment
        self.status = 'pending'
        self.version = None
        self.versions = []

    def deploy(self, new_version: str):
        if self.version is not None:
            self.versions.append(self.version)

        self.version = new_version
        self.status = 'deployed'

    def rollback(self) -> bool:
        if not self.versions:
            return False

        self.version = self.versions.pop()
        self.status = 'rolled_back'
        return True

    def check_status(self) -> dict:
        return {
            'service_name': self.service_name,
            'environment': self.environment,
            'status': self.status,
            'version': self.version
        }
```
