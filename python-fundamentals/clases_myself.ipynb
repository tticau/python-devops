| Pas | Concept Cheie | Ce Construim | Cum se Leagă de Exercițiul Final (Udemy) |
| :--- | :--- | :--- | :--- |
| **Pasul 1** *(Rezolvat)* | **Inițializarea stării** | Atribute de bază în `__init__` (`self.name`, `self.is_running`). | Setează starea de start: `self.status = 'pending'`, `self.version = None`. |
| **Pasul 2** *(Rezolvat)* | **Modificare și raportare** | Metode cu argumente (`update_version`) și export în dicționar (`get_info`). | Baza pentru logica din `deploy()` și dicționarul cerut de `check_status()`. |
| **Pasul 3** *(Curent)* | **Salvarea stării anterioare (`append`)** | O listă-istoric (`self.history`) care reține versiunea veche înainte de suprascriere. | Asigură că înainte de a seta noua versiune în `deploy()`, nu pierzi trecutul. |
| **Pasul 4** *(Următorul)* | **Reversibilitate cu stivă (`pop`)** | Metodă de undo: extragi ultima versiune salvată și verifici dacă lista e goală (`if not history:`). | Mecanismul exact din `rollback()` (scoaterea ultimei versiuni și tratarea eșecului cu `return False`). |
| **Pasul 5** *(Final)* | **Asamblarea completă** | Unificarea tuturor pieselor + mașina de stări (`pending` -> `deployed` -> `rolled_back`). | Rezolvarea autonomă, de la zero, a clasei complete `Deployment`. |

```python
class Deployment:
    """
    Manages the state and version history of a software deployment.
    """

    def __init__(self, service_name: str, environment: str):
        # 1. Salvează service_name și environment
        # 2. Setează self.status inițial pe 'pending'
        # 3. Setează self.version pe None
        # 4. Setează self.versions (istoricul) ca listă goală []
        self.service_name = service_name
        self.environment = environment
        self.status = "pending"
        self.version = None
        self.versions = []

    def deploy(self, new_version: str):
        # 1. Dacă self.version NU este None:
        #    salvează-l în self.versions cu .append()
        # 2. Setează self.version cu new_version
        # 3. Setează self.status pe 'deployed'
        if self.version is not None:
            self.versions.append(self.version)
        self.version = new_version
        self.status = "deployed"

    def rollback(self) -> bool:
        # 1. Verifică dacă istoricul self.versions este gol -> returnează False
        # 2. Extrage ultima versiune cu .pop() și pune-o în self.version
        # 3. Setează self.status pe 'rolled_back'
        # 4. Returnează True
        if not self.versions:
            return False
        self.version = self.versions.pop()
        self.status = "rolled_back"
        return True

    def check_status(self) -> dict:
        # Returnează un dicționar cu cele 4 chei:
        # 'service_name', 'environment', 'status', 'version'
        return {
            'service_name': self.service_name,
            'environment': self.environment,
            'status': self.status,
            'version': self.version
        }
```

```python
d = Deployment("auth-service", "production")
print(d.check_status())
# {'service_name': 'auth-service', 'environment': 'production', 'status': 'pending', 'version': None}

d.deploy("v1.0")
d.deploy("v2.0")
print(d.check_status())
# {'service_name': 'auth-service', 'environment': 'production', 'status': 'deployed', 'version': 'v2.0'}

succes = d.rollback()
print(succes) # True
print(d.check_status())
# {'service_name': 'auth-service', 'environment': 'production', 'status': 'rolled_back', 'version': 'v1.0'}

d.rollback() # revine la None? Nu, versions devine gol după al doilea rollback
succes_2 = d.rollback()
print(succes_2) # False
```
