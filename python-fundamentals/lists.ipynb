# Lists (`list`)

Lists are ordered, mutable sequences defined with square brackets `[]`. You can add, remove, or change items after creation.

## Characteristics and Use Cases
- Ordered: items maintain position
- Mutable: `.append()`, `.insert()`, `.pop()`, `.remove()`
- Ideal for storing sequences where order matters and contents change (e.g., list of servers, deployment steps)

## Accessing Items and Slicing
- Access single elements with `my_list[index]` (0-based). Use negative indices like `my_list[-1]` for the last item.
- Slice with `my_list[start:stop]` to get a sub-list from `start` up to (but not including) `stop`.
- Use three-parameter slicing `my_list[start:stop:step]` for stepping, e.g., `my_list[::2]` selects every other element.
- Omitting `start` or `stop` defaults to the beginning or end of the list respectively, and slicing returns a new list without modifying the original.

```python
def validate_server(server: dict) -> bool:
    """
    Validates a single server dictionary based on a set of rules.

    Rules:
    - Must be a dictionary.
    - Must contain 'name', 'region', and 'status' keys.
    - 'name' and 'region' must be non-empty strings.
    - 'status' must be either 'active' or 'inactive'.

    Args:
        server: A dictionary representing a server.

    Returns:
        True if the server is valid, False otherwise.
    """
    # 1. Check if the input is a dictionary
    if not isinstance(server, dict):
        return False

    # 2. Check for the presence of all required keys
    required_keys = {"name", "region", "status"}
    if not required_keys.issubset(server.keys()):
        return False

    name = server["name"]
    region = server["region"]
    status = server["status"]

    # 3. 'name' and 'region' must be non-empty strings
    if not isinstance(name, str) or not name.strip():
        return False
    if not isinstance(region, str) or not region.strip():
        return False

    # 4. 'status' must be either 'active' or 'inactive'
    if status not in ("active", "inactive"):
        return False

    return True


def generate_inventory_report(servers: list[dict]) -> dict:
    """
    Processes a list of server dictionaries to generate a structured inventory report,
    ignoring any invalid server entries.

    Args:
        servers: A list of dictionaries, where each dictionary represents a server.

    Returns:
        A dictionary where keys are regions, containing 'active' and 'inactive' lists.
    """
    # 1. Handle non-list inputs gracefully
    if not isinstance(servers, list):
        return {}

    # 2. Initialize the empty dictionary for the report
    report = {}

    # 3. Loop through the servers
    for server in servers:
        # 4. Validate server; skip if invalid
        if not validate_server(server):
            continue

        region = server["region"]
        status = server["status"]
        name = server["name"]

        # 5. Initialize the region container if not yet present
        if region not in report:
            report[region] = {
                "active": [],
                "inactive": []
            }

        # 6. Append the server name under the matching status list
        report[region][status].append(name)

    return report

servers =  [

{'name': 'web-01', 'region': 'us-east-1', 'status': 'active'},

{'name': 'db-01', 'region': 'eu-west-1', 'status': 'active'},

{'name': 'app-01', 'region': 'us-east-1', 'status': 'inactive'},

{'name': 'web-02', 'region': 'us-east-1', 'status': 'active'},

{'name': 'monitor-01', 'region': 'eu-west-1', 'status': 'inactive'} # Invalid status

]

generate_inventory_report(servers)
```

```python
def validate_fruit(fruit: dict) -> bool:
    # 1. Check if fruit is a dict
    if not isinstance(fruit, dict):
        return False

    # 2. Check if 'name' and 'in_stock' keys are present
    required_keys = {"name", "in_stock"}
    if not required_keys.issubset(fruit.keys()):
        return False

    name = fruit["name"]

    # 3. Check that fruit['name'] is a non-empty string
    if not isinstance(name, str) or not name.strip():
        return False

    # 4. Check that fruit['in_stock'] is a boolean: isinstance(fruit['in_stock'], bool)
    if not isinstance(fruit["in_stock"], bool):
        return False

    # Return True if all checks pass, else False
    return True


def sort_fruits(fruits: list[dict]) -> dict:
    # 1. Guard check: return default dict if fruits is not a list
    default_report = {
        "in_stock": [],
        "out_of_stock": []
    }

    if not isinstance(fruits, list):
        return default_report

    report = {
        "in_stock": [],
        "out_of_stock": []
    }

    # 2. Loop through fruits
    for fruit in fruits:
        # 3. Skip if not valid
        if not validate_fruit(fruit):
            continue
        # 4. Append fruit['name'] to the right list based on fruit['in_stock']
        if fruit["in_stock"]:
            report["in_stock"].append(fruit["name"])
        else:
            report["out_of_stock"].append(fruit["name"])

    # 5. Return report
    return report


# --- Test it with this sample data ---
sample_fruits = [
    {"name": "Apple", "in_stock": True},
    {"name": "Banana", "in_stock": False},
    {"name": "   ", "in_stock": True},       # Invalid: name is whitespace
    {"name": "Orange"},                      # Invalid: missing 'in_stock' key
    {"name": "Mango", "in_stock": "yes"},    # Invalid: 'in_stock' is not a bool
    {"name": "Pear", "in_stock": True},
]

print(sort_fruits(sample_fruits))
# Expected output:
# {'in_stock': ['Apple', 'Pear'], 'out_of_stock': ['Banana']}
```
